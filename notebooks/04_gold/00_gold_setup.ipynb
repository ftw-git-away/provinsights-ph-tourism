# Gold Control Setup

Creates the Gold schema and the run log used by the Gold notebooks.

| Object | Purpose |
| --- | --- |
| `tuklas_dev`.`04_gold` | Star schema: `dim_province`, `dim_year`, `fact_province_year` |
| `04_gold`.`gold_run_log` | One row per Gold notebook run |

Data-quality results go to the existing `06_data_quality`.`dq_results`.

**Idempotency:** only `CREATE ... IF NOT EXISTS`, so it is safe to rerun.

```python
CATALOG = "tuklas_dev"
GOLD_SCHEMA = "04_gold"
DQ_SCHEMA = "06_data_quality"

spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{CATALOG}`.`{GOLD_SCHEMA}`")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{CATALOG}`.`{DQ_SCHEMA}`")

spark.sql(f"""
CREATE TABLE IF NOT EXISTS `{CATALOG}`.`{GOLD_SCHEMA}`.`gold_run_log` (
    run_id             STRING,
    notebook_name      STRING,
    started_at         TIMESTAMP,
    completed_at       TIMESTAMP,
    tables_written     BIGINT,
    rows_written       BIGINT,
    dq_checks_run      BIGINT,
    dq_block_failures  BIGINT,
    dq_review_flags    BIGINT,
    status             STRING,
    notes              STRING
) USING DELTA
""")
print("✓ Gold schema and gold_run_log created / verified.")
```
